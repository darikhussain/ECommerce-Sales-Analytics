```sql
SELECT AVG(Total_Spent)
FROM (SELECT
    CustomerID,
    SUM(Amount) AS Total_Spent
FROM Orders
GROUP BY CustomerID
) AS Customer_Total;
```

```sql
SELECT
    c.CustomerName,
    SUM(o.Amount) AS Total_Spent
FROM Customers c
INNER JOIN Orders o
ON c.CustomerID = o.CustomerID
GROUP BY c.CustomerID, c.CustomerName
HAVING SUM(o.Amount) > (
    SELECT AVG(Total_Spent)
    FROM(
        SELECT
            CustomerID,
            SUM(Amount) AS Total_Spent
            FROM Orders
            GROUP BY CustomerID
    ) AS Customer_Totals
);
```
